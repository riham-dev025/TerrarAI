```python
import os

data_dir = "../data/raw/trashnet/data/dataset-resized"  #directory of the dataset
classes = sorted(os.listdir(data_dir)) #listdir: grab everything from the folder. sorted() :sort alphabateically

for c in classes:
    class_path = os.path.join(data_dir,c)
    if os.path.isdir(class_path):
        n = len(os.listdir(class_path))
        print(f"{c}: {n} images")

## Class Distribution
- cardboard: 403
- glass: 501
- metal: 410
- paper: 594
- plastic: 482
- trash: 137

**Finding:** trash is significantly underrepresented (~3.6-4.3x smaller than other classes).
This will need addressing via class weighting or oversampling during training (Phase 2).
```
